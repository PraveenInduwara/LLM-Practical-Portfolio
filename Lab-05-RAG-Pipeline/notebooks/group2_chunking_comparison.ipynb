# Lab 5 Group 2: Context-aware RAG comparison

This notebook explores the six-page Lab 5 handout and prepares the three chunking strategies. The measured handout run is recorded in `evaluation/results/`. A separate measured run uses an eight-question dataset and the first 50 pages of the official 210-page Sri Lankan UGC admissions handbook; its metrics are in `evaluation/results/handbook_sample/`.

The full 210-page handbook could not be indexed within a practical CPU runtime. The 50-page sample is a limited long-document test, not a full-book result. Use the README commands to reproduce each run.

```python
from pathlib import Path
import json
from rag_app.core import load_pdf_directory, make_chunks

CORPUS = Path('data/evaluation_corpus')
pages = load_pdf_directory(CORPUS)
print(f'{len(pages)} nonempty PDF pages')
for page in pages[:3]:
    print(f"\n{page.metadata['source']} p.{page.metadata['page']}")
    print(page.page_content[:700])
```

```python
for strategy in ('fixed', 'sections', 'parent_child'):
    chunks, parents, _ = make_chunks(pages, strategy)
    print(f'{strategy:14s} chunks={len(chunks):3d}, parent passages={len(parents):3d}')

cases = json.loads(Path('evaluation/dataset.json').read_text(encoding='utf-8'))
print(f'\nEvaluation questions: {len(cases)}')
print(cases[0]['question'])
```
